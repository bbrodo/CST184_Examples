```python
# Create mock data
set.seed(123)

co2 <- runif(50, min = 5, max = 20)  # CO₂ emissions per capita (in tonnes)
temperature_anomaly <- 0.3 * co2 + rnorm(50, mean = 0, sd = 1.5)  # Some random noise

climate_df <- data.frame(
  co2,
  temperature_anomaly
)

head(climate_df)
```

```python
library(ggplot2)

ggplot(climate_df, aes(x = co2, y = temperature_anomaly)) +
  geom_point(color = "darkblue") +
  labs(
    title = "CO₂ Emissions vs. Temperature Anomaly",
    x = "CO₂ Emissions (tonnes per capita)",
    y = "Temperature Anomaly (°C)"
  ) +
  theme_minimal()
```

```python
# Fit the model
model <- lm(temperature_anomaly ~ co2, data = climate_df)

# Display model summary
summary(model)
```

```python
ggplot(climate_df, aes(x = co2, y = temperature_anomaly)) +
  geom_point(color = "darkblue") +
  geom_smooth(method = "lm", se = TRUE, color = "red") +
  labs(
    title = "Regression: CO₂ vs. Temperature Anomaly",
    x = "CO₂ Emissions (tonnes per capita)",
    y = "Temperature Anomaly (°C)"
  ) +
  theme_minimal()
```

Slope is positive. <br>
R squared values tells me that around 60% of the Temperature anomaly is explained by the Co2 emissions. <br>
The relationship is very statistically significant as 1.822e-11 is well below 0.05. <br>
If Co2 emission increased by 5 tones per capita the model predicts that the temperature would increase by the slope * 5. So 0.33817 * 5 = 1.69085°C increase
