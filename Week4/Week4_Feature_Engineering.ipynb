```python

# Week 4 – Categorical Feature Engineering Activity

import pandas as pd
from itertools import product

# Step 1: Define regions and energy types
regions = ['Asia', 'Africa']
energy_types = ['Solar', 'Wind', 'Hydro', 'Geothermal']
time = ['day', 'night']

# Step 2: Create Cartesian product (all region-energy combinations)
combos = list(product(regions, energy_types, time))

# Step 3: Create a DataFrame
df = pd.DataFrame(combos, columns=['region', 'energy_type', 'time'])

# Step 4: Create a combined feature
df['region_energy_time'] = df['region'] + "_" + df['energy_type'] + "_" + df['time']

# Step 5: Display the resulting DataFrame
print("Engineered Feature Table:")
print(df)
```

Combinations scale by region x energy_type

Adding more categories just multiplies the total combinations by the amount of elements in the category

If one category had 10 values there would be 10 times more combinations. These are silly questions, its literally just the product of each category.
