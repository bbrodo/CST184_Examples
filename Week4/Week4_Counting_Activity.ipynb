```python

# Week 4 – Counting Principles in Python

import math
from itertools import permutations, combinations

# Step 1: Factorial – Total ways to arrange 5 regions
print("Step 1 – Factorial:")
print("4! =", math.factorial(4))

# Step 2: Permutations – How many ways to apply 3 out of 4 regions (order matters)
print("\nStep 2 – Permutations (nPr):")
n, r = 4, 3
perm_result = math.perm(n, r)  # Requires Python 3.8+
print(f"Permutations of {n} items taken {r} at a time: {perm_result}")

# Step 3: Combinations – How many ways to choose 3 regions (order doesn't matter)
print("\nStep 3 – Combinations (nCr):")
comb_result = math.comb(n, r)
print(f"Combinations of {n} items taken {r} at a time: {comb_result}")

# Step 4: List actual combinations
regions = ['Africa', 'Asia', 'Europe', 'North America', 'Oceania', 'South America']
print("\nAll 2-region combinations (unordered):")
for combo in combinations(regions, 2):
    print(combo)

# Step 5: List actual permutations
print("\nAll 2-region permutations (ordered):")
for perm in permutations(regions, 2):
    print(perm)
    break  # remove this line to view all permutations
```

Changing the number of regions from 5 to 4 causes the permutations of items taken 3 at a time to be the same as 4!
