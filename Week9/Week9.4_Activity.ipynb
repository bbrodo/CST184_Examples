```python
from networkx.algorithms.bipartite.basic import density

install.packages("ggplot2")
library(ggplot2)
```

```python
n <- 10
p <- 0.7
x <- 0:n
y <- dbinom(x, size = n, prob = p)

binom_df <- data.frame(
  successes = x,
  probability = y)

# Plot
ggplot(binom_df, aes(x = successes, y = probability)) +
  geom_bar(stat = "identity", fill = "steelblue") +
  geom_vline(xintercept = (n * p), linetype = "dashed", color = "red") +
  labs(
    title = "Binomial Distribution with Expected Value",
    x = "Number of People Saying Yes",
    y = "Probability"
  ) +
  theme_minimal()
```

```python
lambda <- 4
x <- 0:15
y <- dpois(x, lambda = lambda)

poisson_df <- data.frame(
  events = x,
  probability = y
)

ggplot(poisson_df, aes(x = events, y = probability)) +
  geom_bar(stat = "identity", fill = "firebrick") +
  labs(
    title = "Poisson Distribution (λ = 4)",
    x = "Number of Events",
    y = "Probability"
  ) +
  theme_minimal()
```

```python
mean <- 2300
sd <- 400
x <- seq(1000, 3500, length.out = 200)
y <- dnorm(x, mean = mean, sd = sd)
df <- data.frame(x,y)

ggplot(df, aes(x, y)) +
  geom_line(color = "firebrick", size = 1.2) +
  labs(
    title = "Average Daily Calorie Intake)",
    x = "Calorie Intake (kcal)",
    y = "Density"
  ) +
  theme_minimal()
```
