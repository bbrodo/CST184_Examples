```python
import pandas as pd

df = pd.read_csv("week6_grouping_filtering.csv")
print(df.head())
```

```python
# Check for missing values
print(df.isnull().sum())
print(len(df))
# Drop rows with missing attendance
df_cleaned = df.dropna(subset=['Attendance'])
print("Rows after cleaning:", len(df_cleaned))
```

5 rows were removed, this cleanses the data and makes it more reliable

```python
df_2023 = df_cleaned[df_cleaned['Year'] == 2023]
print(df_2023.head())
print("Rows remaining" , len(df_2023))
```

58 rows remaining for 2023

```python
summary = df_2023.groupby(['Region', 'Gender'])['Attendance'].agg(['mean', 'count']).reset_index()
print(summary)
pivot = df_2023.pivot_table(values='Attendance', index='Region', columns='Gender', aggfunc='mean')
print(pivot)
```

```python
region_avg = df_2023.groupby('Region')['Attendance'].mean()
high_regions = region_avg[region_avg > 85]
print(high_regions)
```

This filter provides no results as no regions have greater than 85% attendance

```python
result = (
    df_cleaned[df_cleaned['Year'] == 2023]
    .groupby(['Region', 'Gender'])['Attendance']
    .agg(['mean', 'count'])
    .reset_index()
)
print(result)
```
