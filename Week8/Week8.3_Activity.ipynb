```python
climate <- matrix(c(
  9.2, 1.1, 0.8,
  6.5, 0.9, 1.2,
  2.3, 0.7, 2.5,
  3.8, 1.0, 1.5,
  7.1, 1.2, 0.9
), nrow=5, byrow=TRUE)

colnames(climate) <- c("CO2", "Temp", "Deforestation")
rownames(climate) <- c("A", "B", "C", "D", "E")

print(climate)
```

```python
scaled_climate <- scale(climate)
print(scaled_climate)
```

```python
cor_matrix <- cor(scaled_climate)
print(cor_matrix)
```

1
Strongest positive correlation: CO2 and Temp (0.749). <br>
2
Yes: Deforestation is negatively correlated with both CO2 (-0.920) and Temp (-0.898).<br>
3
Weak/no correlation: none; all correlations are fairly strong.<br>
In this small dataset, higher CO2 tends to go with higher temperature. Deforestation moves in the opposite direction from both CO2 and temperature.

```python
climate <- rbind(climate, c(10.0, 1.5, 0.4))
rownames(climate)[6] <- "F"

scaled_climate <- scale(climate)
cor_matrix <- cor(scaled_climate)
print(cor_matrix)
```

Correlations get stronger when adding a sixth country, this happens because the new country fits the existing pattern very strongly: high CO2 and temp, and low deforestation.
