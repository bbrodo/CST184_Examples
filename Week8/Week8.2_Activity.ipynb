```python
# Create a 6x3 matrix: [CO2, Temp Anomaly, Deforestation Rate]
climate <- matrix(c(
  9.2, 1.1, 0.8,   # Country A
  6.5, 0.9, 1.2,   # Country B
  2.3, 0.7, 2.5,   # Country C
  3.8, 1.0, 1.5,   # Country D
  7.1, 1.2, 0.9,   # Country E
  5.4, 0.8, 1.8    # Country F
), nrow=6, byrow=TRUE)

colnames(climate) <- c("CO2", "Temp", "Deforestation")
rownames(climate) <- c("A", "B", "C", "D", "E", "F")

print(climate)
```

```python
scaled_climate <- scale(climate)
print(scaled_climate)
```

```python
flip <- matrix(c(2, 0, 0,
                 0, 1, 0,
                 0, 0, -1), nrow=3,
               dimnames=list(colnames(scaled_climate), colnames(scaled_climate)))

transformed <- scaled_climate %*% flip
print(transformed)
```

Doubling CO2 weights emissions more heavily, while inverting deforestation still makes higher deforestation reduce the score.

```python
transposed <- t(transformed)
print(transposed)
```

```python
original_flip <- matrix(c(1, 0, 0,
                          0, 1, 0,
                          0, 0, -1), nrow=3,
                        dimnames=list(colnames(scaled_climate), colnames(scaled_climate)))

original_transposed <- t(scaled_climate %*% original_flip)
comparison <- transposed - original_transposed

print(comparison)
```
